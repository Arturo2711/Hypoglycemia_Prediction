```python
import pandas as pd 
import numpy as np 
import os 
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

def optimize_ARIMA(endog: pd. Series, order_list: list, d: int) -> pd.DataFrame:
    results = []

    for order in order_list:
        try:
            model = SARIMAX(endog, order=(order [0], d, order[1]), simple_differencing=False).fit(disp=False)
        except:
            continue
        aic = model.aic
        bic = model.bic
        results.append([order, round(aic, 2), round(bic, 2)])

    result_df = pd.DataFrame (results)
    result_df.columns = ['(p,q)', 'AIC', 'BIC']
    #Sort in ascending order, lower AIC is better
    result_df = result_df.sort_values(by='AIC', ascending=True). reset_index(drop=True)
    result_2_df = result_df.sort_values(by='BIC', ascending=True). reset_index(drop=True)
    return result_df[['(p,q)', 'AIC']], result_2_df[['(p,q)', 'BIC']]
```

```python
from itertools import product

range_start = 0
range_end = 5
combinations = list(product(range(range_start, range_end), repeat=2))
```

```python
data_dir = '../Data/Preprocessed'
base_paths = os.listdir(data_dir)
full_paths = [os.path.join(data_dir, p) for p in base_paths]
orders = {}
for i, fp in enumerate(full_paths):
    key = base_paths[i].split('.')[0]
    data = pd.read_csv(fp, sep=';')['glucose'][:2290]
    aic, bic = optimize_ARIMA(endog=data, order_list=combinations, d=1)
    orders[key] = aic.iloc[0]['(p,q)']
    print(key)
    print(aic.iloc[0])
    print(bic.iloc[0])
    print('*'*20)
    
```
