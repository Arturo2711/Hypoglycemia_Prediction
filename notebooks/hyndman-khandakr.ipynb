```python
import os
import pandas as pd
import pmdarima as pm
from pmdarima import model_selection
from sklearn.metrics import mean_squared_error
import numpy as np
import warnings
import time
import matplotlib.pyplot as plt
import seaborn as sns
```

# Hyndman-Khandakar algorithm
1. Determine d using repeated stationarity tests.
2. choose p and q, by minimizing the AICcs.

```python
warnings.filterwarnings("ignore")

def hyndman_khandakar_non_seasonal_forecast(time_series: pd.Series, dif : int, p: int, q: int):

    print(f"Arima (Hyndman-Khandakar logic)")
    
    model = pm.auto_arima(
        y = time_series,
        start_p = 1, start_q = 1,
        max_p = p, max_q = q,
        seasonal = False,
        maxiter = 15,
        d = dif,
        test = "kpss",
        stepwise = True,
        trace = False,
        error_action = 'trace',
        suppress_warnings = True,
        information_criterion = 'aicc'
    )

    print(f"\nBest Model")
    print(model.summary())
    print("-" * 40)
    print(f"Selected ARIMA Order: {model.order}")

    return model
```

```python
def forecasting_HK(model, h):
    forecasts = model.predict(h)
    
    if isinstance(time_series, pd.Series):
        try:
            last_date = time_series.index[-1]
            if isinstance(last_date, pd.Timestamp):
                 # For datetime index
                freq = time_series.index.freq if time_series.index.freq is not None else pd.infer_freq(time_series.index)
                forecast_index = pd.date_range(start=last_date, periods=h + 1, freq=freq)[1:]
            else:
                 # For a simple integer index
                forecast_index = pd.RangeIndex(start=len(time_series), stop=len(time_series) + h)
        except Exception:
            forecast_index = pd.RangeIndex(start=len(time_series), stop=len(time_series) + h)


    forecast_series = pd.Series(forecasts, index=forecast_index, name='Forecast')
    print(f"forecast series : {forecast_series}")
    return forecast_series
```

```python
def cast_to_time(df):
    df['time'] = pd.to_datetime(df['time'])
    return df
```

```python
data_dir = '../Data/Preprocessed'
base_paths = os.listdir(data_dir)
full_paths = [os.path.join(data_dir, p) for p in base_paths]
patients = {base_paths[i].split('.')[0] : cast_to_time(pd.read_csv(fp, sep=';')) for i, fp in enumerate(full_paths)}
ids = list(patients.keys())
s = 1.5
```

```python
too_long = ["HUPA0027P", "HUPA0028P", "HUPA0026P"]
for i in too_long:
    ids.remove(i)
```

## no differentiation, d = 0

```python
results_d0 = []

for i in ids:
    d = 0
    ts = patients[i].loc[:, "glucose"]
    print(f"longitud : {len(ts)}")
    print(f"paciente : {i}")
    # Run the non-seasonal HK algorithm
    result = hyndman_khandakar_non_seasonal_forecast(time_series = ts, dif = d, p = 6, q = 6)
    results_d0.append(result)
    time.sleep(s)
```

## Differentiation = 1

```python
results_d1 = []

for i in ids:
    d = 1
    ts = patients[i].loc[:, "glucose"]
    print(f"longitud : {len(ts)}")
    print(f"paciente : {i}")
    # Run the non-seasonal HK algorithm
    result = hyndman_khandakar_non_seasonal_forecast(time_series = ts, dif = d, p = 6, q = 6)
    results_d1.append(result)
    time.sleep(s)
```

## Hyperparameter Analysis

```python
def hyperparameters(results_dn):
    d_p = []
    d_q = []
    for r_dn in results_dn:
        p = r_dn.order[0]
        q = r_dn.order[2]
        d_p.append(p)
        d_q.append(q)
    return d_p, d_q
```

```python
d0_p, d0_q = hyperparameters(results_d0)
aicc_d0 = [res.aicc() for res in results_d0]
d1_p, d1_q = hyperparameters(results_d1)
aicc_d1 = [res.aicc() for res in results_d1]
```

```python
def unit_circle(ar_roots: np.ndarray, ma_roots: np.ndarray, title: str = "ARIMA Characteristic Roots on the Unit Circle"):
    fig, ax = plt.subplots(figsize=(8, 8))
    all_roots = []
    max_modulus = 1.0

    # Ensure inputs are treated as complex numbers and collect all roots for limits calculation
    def process_roots(roots, name):
        nonlocal max_modulus
        if roots is None or len(roots) == 0:
            return np.array([], dtype=complex)
        
        if not roots.dtype == complex:
            print(f"Warning: Input array for {name} roots was converted to complex.")
            roots = roots.astype(complex)
        
        all_roots.extend(list(roots))
        if len(roots) > 0:
            max_modulus = max(max_modulus, np.max(np.abs(roots)))
        return roots

    ar_roots = process_roots(ar_roots, 'AR')
    ma_roots = process_roots(ma_roots, 'MA')
    
    # Ensure inputs are treated as complex numbers and collect all roots for limits calculation
    # If no roots are provided after processing, just return the empty plot
    if not all_roots:
        print("No AR or MA roots provided to plot.")
        ax.set_title(title)
        plt.show()
        return

    # --- 1. Draw the Unit Circle ---
    circle_theta = np.linspace(0, 2 * np.pi, 100)
    circle_x = np.cos(circle_theta)
    circle_y = np.sin(circle_theta)
    ax.plot(circle_x, circle_y, 'k--', alpha=0.5, label='Unit Circle (Modulus=1)')

    # --- 2. Helper function for plotting a set of roots ---
    def plot_roots_set(roots, root_type, marker_stable, marker_unstable, color_stable, color_unstable):
        if len(roots) == 0:
            return

        modulus = np.abs(roots)
        inside_mask = modulus < 1
        outside_mask = modulus > 1
        on_circle_mask = np.isclose(modulus, 1)

        # Plot Outside Roots (Stable/Invertible)
        ax.scatter(roots[outside_mask].real, roots[outside_mask].imag,
                   c=color_stable, marker=marker_stable, s=60, 
                   label=f'{root_type} Outside ({np.sum(outside_mask)})', zorder=5)

        # Plot Inside Roots (Non-stable/Non-invertible - Problematic)
        ax.scatter(roots[inside_mask].real, roots[inside_mask].imag,
                   c=color_unstable, marker=marker_unstable, s=120, 
                   label=f'{root_type} Inside ({np.sum(inside_mask)})', zorder=5)
        
        # Plot On Circle Roots (Unit/Critical Root)
        if np.sum(on_circle_mask) > 0:
             ax.scatter(roots[on_circle_mask].real, roots[on_circle_mask].imag,
                       c='green', marker='D', s=70, 
                       label=f'{root_type} On Circle ({np.sum(on_circle_mask)})', zorder=6)

    # Plot AR Roots (Blue/Red)
    plot_roots_set(ar_roots, 'AR', 'o', 'x', 'blue', 'red')

    # Plot MA Roots (Purple/Orange)
    plot_roots_set(ma_roots, 'MA', 's', '^', 'purple', 'orange')

    # --- 3. Axes and Labels ---
    
    # Draw real and imaginary axes
    ax.axhline(0, color='gray', linestyle='-', linewidth=0.5)
    ax.axvline(0, color='gray', linestyle='-', linewidth=0.5)

    # Set limits and aspect ratio
    limit = max(1.5, max_modulus * 1.1)
    ax.set_xlim(-limit, limit)
    ax.set_ylim(-limit, limit)
    ax.set_aspect('equal', adjustable='box')

    # Add labels and title
    ax.set_xlabel('Real Axis')
    ax.set_ylabel('Imaginary Axis')
    ax.set_title(title, fontsize=14)
    ax.legend(loc='best', fontsize=9)
    ax.grid(True, linestyle=':', alpha=0.6)

    plt.show()
```

```python
def expected_value(df, col):
    probs = df[col].value_counts(normalize = True)
    product = probs * probs.index
    return product.sum()
```
